```python
import os

folders = [
    "data/raw",
    "data/processed",
    "notebooks",
    "src",
    "models",
    "results"
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)

    # Crée un fichier vide pour conserver les dossiers vides dans Git
    if folder in ["data/raw", "data/processed", "models", "results"]:
        with open(os.path.join(folder, ".gitkeep"), "w") as f:
            pass

print("Project architecture created!")
```

```python
files = [
    "src/__init__.py",
    "src/preprocessing.py",
    "src/features.py",
    "src/embeddings.py",
    "src/similarity.py",
    "src/evaluation.py",
    "app.py",
    "requirements.txt",
    "README.md",
    ".gitignore"
]

for file in files:
    if not os.path.exists(file):
        with open(file, "w") as f:
            pass
```
